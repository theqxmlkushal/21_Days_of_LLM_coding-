```python
#Python script using numpy that simulates a single neuron. It must take 3 inputs, multiply them by 3 weights, add a bias, and output one number.
#Code: Manual Dot Product vs. np.dot.
```

```python
import numpy as np

def neuron_layer_1():
    inputs = np.array([], dtype=int)

    while inputs.size < 3:
        value = int(input("Enter input number: "))
        inputs = np.append(inputs, value)

    weights = np.array([], dtype=float)
    while weights.size < 3:
        value = int(input("Enter weights number: "))
        weights = np.append(weights, value)
        
    bias = input("Enter the bias: ",end="\n")

    #manual dot product
    output = inputs[0] * weights[0] + inputs[1] * weights[1] + inputs[2] * weights[2] + bias
    print("Manual dot product output:", output)

    #dot product using numpy
    output = np.dot(inputs,weights) + bias
    print("dot product using numpy output:",output)

    return output
```

```python
#activation function     
def making_activation_func(output,choice):
    choice = str()
    if choice == "relu":
        relu = np.maximum(0, output)
        return relu
    elif choice == "sigmoid":
        sigmoid = 1/(1 + np.exp(-output))
        return sigmoid
    elif choice == "softmax":
        softmax = np.exp(output - np.max(output)) / np.sum(np.exp(output - np.max(output)))
        return softmax
    else:
        print("Enter a valid choice sorry :(")
        return 0
```
